# Detección de Outliers con DBSCAN

Proyecto de clustering density-based listo para **Google Colab**.

**Pregunta:** ¿cómo identificar outliers/anomalías con DBSCAN?

**Método:** DBSCAN (Density-Based Spatial Clustering).

**Stack:** pandas, seaborn, scikit-learn

Dataset simulado con outliers.

## 0. Setup

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_blobs

sns.set_theme(style="whitegrid")
RANDOM_STATE = 42
print("OK")
```

## 1. Generar datos con outliers

```python
np.random.seed(RANDOM_STATE)
# Clusters normales
X_normal, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.8, random_state=RANDOM_STATE)
# Outliers
X_outliers = np.random.uniform(low=-10, high=10, size=(20, 2))
X = np.vstack([X_normal, X_outliers])
print(X.shape)
```

## 2. Visualización inicial

```python
plt.figure(figsize=(8, 5))
plt.scatter(X[:,0], X[:,1], s=20, alpha=0.6)
plt.title("Datos con outliers")
plt.show()
```

## 3. Preprocesamiento

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled.shape)
```

## 4. DBSCAN

```python
dbscan = DBSCAN(eps=0.5, min_samples=5)
labels = dbscan.fit_predict(X_scaled)
print("Clusters encontrados:", len(set(labels)) - (1 if -1 in labels else 0))
print("Outliers (label -1):", np.sum(labels == -1))
```

## 5. Visualización de resultados

```python
plt.figure(figsize=(8, 5))
unique_labels = set(labels)
colors = plt.cm.tab10(np.linspace(0, 1, len(unique_labels)))
for k, col in zip(unique_labels, colors):
    if k == -1:
        col = 'red'
        marker = 'x'
    else:
        marker = 'o'
    class_member_mask = (labels == k)
    plt.scatter(X[class_member_mask, 0], X[class_member_mask, 1],
                c=[col], marker=marker, s=40, label=f'Cluster {k}' if k != -1 else 'Outliers')
plt.title("DBSCAN: clusters y outliers")
plt.legend()
plt.show()
```

## 6. Ajuste de parámetros

```python
# Probar distintos eps
for eps in [0.3, 0.5, 0.8]:
    db = DBSCAN(eps=eps, min_samples=5)
    labs = db.fit_predict(X_scaled)
    n_out = np.sum(labs == -1)
    n_clust = len(set(labs)) - (1 if -1 in labs else 0)
    print(f'eps={eps}: clusters={n_clust}, outliers={n_out}')
```

## 7. Conclusiones

- DBSCAN identifica clusters de densidad arbitraria y marca outliers como -1.
- No requiere especificar el número de clusters.
- Ideal para detección de anomalías.

**Consejo:** ajustar `eps` y `min_samples` según la escala de los datos.
